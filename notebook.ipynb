## 2. Data Wrangling

### Definition

**Data wrangling**, also known as **data munging**, is the process of cleaning, transforming, and preparing raw data for analysis.

### Key Activities

- **Handling Missing Values:** Identifying and treating missing or null values in the dataset.
- **Removing Duplicates:** Detecting and eliminating duplicate records to maintain data accuracy.
- **Addressing Inconsistencies:** Correcting inconsistent values, formats, and data types.
- **Data Transformation:** Converting data into suitable formats and creating new features when required.
- **Data Validation:** Checking data quality, accuracy, and consistency before analysis.

### Objective

The primary objective of data wrangling is to ensure that the dataset is **clean, consistent, reliable, and ready for exploratory data analysis (EDA) and further modeling**.

### 2.4 Data Cleaning & Dropping Insignificant Features


```python
# Showing Random Sample of Data.
df.sample(5)
```

### Inference

- **`RowNumber` and `CustomerId`:** These columns serve as identifiers for individual records or customers. They do not provide meaningful information for predicting customer churn and can generally be excluded from the analysis.
- **`Surname`:** This column has high cardinality, meaning it contains many distinct values. Since customer surnames are unlikely to provide meaningful predictive information, this feature can also be excluded from the model.

**Conclusion:** We can drop `RowNumber`, `CustomerId`, and `Surname` from the dataset before proceeding with further exploratory data analysis and predictive modeling.

```python
# Dropping Insignificant Features.
df.drop(columns=['RowNumber', 'CustomerId', 'Surname'], inplace=True)
```
